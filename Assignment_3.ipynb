```python
data_emp = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp_df = pd.DataFrame(data_emp)
print(emp_df.head())
```

```python
print("a) Shape: ", emp_df.shape)
print("\nb) Info: ", emp_df.info())
print("\nc) Descriptive Statistics: ", emp_df.describe())
print("\nd)", emp_df.head(5))
print(emp_df.tail(3))
print("\ne) Average Salary:", emp_df['Salary'].mean())
print("Total Bonus:", emp_df['Bonus'].sum())
print("Youngest Employee Age:", emp_df['Age'].min())
print("Highest Performance Rating:", emp_df['Rating'].max())
sorted_emp = emp_df.sort_values(by='Salary', ascending=False)
print("\nf)", sorted_emp)
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'

emp_df['Performance_Category'] = emp_df['Rating'].apply(categorize_rating)
print("\nh) Missing values per column:\n", emp_df.isnull().sum())
emp_df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
filtered_emp = emp_df[(emp_df['Years_of_Experience'] > 5) & (emp_df['Department'] == 'IT')]
print("\nj)")
print(filtered_emp)
emp_df['Tax'] = emp_df['Salary'] * 0.10
emp_df.to_csv('modified_employees.csv', index=False)
print("\nl) File 'modified_employees.csv'  saved")
```
