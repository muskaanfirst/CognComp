```python
import random
roll_no = 1024170248
random.seed(roll_no)
numbers = [random.randint(100,900) for i in range(100)]
print("i.", numbers)
```

```python
odd = [x for x in numbers if x % 2 != 0]
print("ii.", len(odd), odd)
```

```python
even = [x for x in numbers if x % 2 == 0]
print("ii.", len(even), even)
```

```python
def prime(n):
  if n <= 1:
    return False
  for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
      return False
  return True
prime_num = [x for x in numbers if prime(x)]
print("iii.", len(prime_num), prime_num)
```

```python
from collections import Counter
most_common_num, count = Counter(numbers).most_common(1)[0]
print(f"v. Most frequent number: {most_common_num} (occurs {count} times)")
```

```python
roll_no = (1,0,2,4,1,7,0,2)
A = {int(digit)*7 for digit in roll_no}
B = {int(digit)*9 for digit in roll_no}
print("A = ", A)
print("B = ", B)
```

```python
print("vi.", A.union(B))
```

```python
print("vii.", A.intersection(B))
```

```python
print("viii.", A.difference(B))
print("viii.", B.difference(A))
```

```python
print("ix.", A.symmetric_difference(B))
```

```python
print("x.", A.issubset(B))
print("x.", B.issuperset(A))
```

```python
X = int(input(" "))
A.discard(X)
print("xi.", A)
```
