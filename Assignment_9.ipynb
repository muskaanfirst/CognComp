```python
import string
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources (run once)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Original 6-sentence paragraph
text = (
    "Artificial Intelligence isn't just a futuristic concept anymore; it is transforming modern healthcare and robotics. "
    "In 2026, researchers deployed state-of-the-art neural networks achieving 98.5 percent accuracy in early diagnostics. "
    "Many engineers believe computing constants like 3.14 remain foundational to algorithm design. "
    "However, ethical deployment requires strict oversight so technology doesn't compromise user privacy. "
    "Students exploring Cognitive Computing can build real-world prototypes using Python libraries. "
    "For collaboration, contact lab@cognitive-ai.org, call +91 9876543210 or 123-456-7890, or visit https://www.cognitive-ai.org."
)

# 1. Convert text to lowercase
lower_text = text.lower()

# Sentence tokenization is performed BEFORE removing punctuation so sentence boundaries (.) are preserved
sentences = sent_tokenize(lower_text)

# Remove punctuation
clean_text = lower_text.translate(str.maketrans('', '', string.punctuation))

# 2. Word tokenization
words = word_tokenize(clean_text)

# 3. Remove stopwords using NLTK's English stopwords list
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

# 4. Word frequency distribution (excluding stopwords)
freq_dist = FreqDist(filtered_words)

print("Sentences:", sentences)
print("\nWord Tokens (without punctuation):", words)
print("\nFiltered Words (without stopwords):", filtered_words)
print("\nTop 10 Word Frequencies:")
for word, count in freq_dist.most_common(10):
    print(f"{word}: {count}")
```

```python
from nltk.stem import PorterStemmer, LancasterStemmer, WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize stemmers and lemmatizer
porter = PorterStemmer()
lancaster = LancasterStemmer()
lemmatizer = WordNetLemmatizer()

# 1–3. Apply PorterStemmer, LancasterStemmer, and WordNetLemmatizer on Q1 filtered_words
comparison_data = []
for word in filtered_words:
    p_stem = porter.stem(word)
    l_stem = lancaster.stem(word)
    lemma = lemmatizer.lemmatize(word)
    comparison_data.append((word, p_stem, l_stem, lemma))

# 4. Compare and display results
print(f"{'Original Word':<18} {'Porter Stem':<16} {'Lancaster Stem':<16} {'Lemma':<16}")
print("-" * 68)
for orig, p, l, lem in comparison_data:
    print(f"{orig:<18} {p:<16} {l:<16} {lem:<16}")
```

```python
import re

# 1. Original text from Q1
# 2a. Extract all words with more than 5 letters
words_more_than_5 = re.findall(r'\b[a-zA-Z]{6,}\b', text)

# 2b. Extract all numbers (integers and decimals)
numbers = re.findall(r'\b\d+(?:\.\d+)?\b', text)

# 2c. Extract all capitalized words
capitalized_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)

# 3a. Split text into words containing only alphabets (removing digits & special characters)
alpha_words = [word for word in re.split(r'[^a-zA-Z]+', text) if word]

# 3b. Extract words starting with a vowel
vowel_words = [word for word in alpha_words if word[0].lower() in 'aeiou']

print("2a. Words with > 5 letters:\n", words_more_than_5)
print("\n2b. Extracted numbers:\n", numbers)
print("\n2c. Capitalized words:\n", capitalized_words)
print("\n3a. Alphabet-only words (via re.split):\n", alpha_words)
print("\n3b. Words starting with a vowel:\n", vowel_words)
```

```python
import re

# 2. Custom tokenization function
def custom_tokenize(input_text):
    """
    - Keeps contractions intact (e.g., "isn't")
    - Keeps hyphenated words intact (e.g., "state-of-the-art")
    - Keeps integers and decimal numbers intact (e.g., "3.14", "98.5")
    - Removes standalone punctuation and special symbols
    """
    pattern = r"\d+(?:\.\d+)?|[a-zA-Z]+(?:[-'][a-zA-Z]+)*"
    return re.findall(pattern, input_text)

custom_tokens = custom_tokenize(text)
print("Custom Tokens:\n", custom_tokens)

# 3. Regex Substitutions (re.sub)
# 3a. Replace email addresses with '<EMAIL>'
cleaned_text = re.sub(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', '<EMAIL>', text)

# 3b. Replace URLs with '<URL>' (preserving trailing sentence punctuation)
cleaned_text = re.sub(r'(?:https?://|www\.)[^\s]+?(?=[.,;!?]?(?:\s|$))', '<URL>', cleaned_text)

# 3c. Replace phone numbers (formats: 123-456-7890 or +91 9876543210) with '<PHONE>'
cleaned_text = re.sub(r'(?:\+91\s\d{10}|\b\d{3}-\d{3}-\d{4}\b)', '<PHONE>', cleaned_text)

print("\nText after Regex Substitutions:\n", cleaned_text)
```
