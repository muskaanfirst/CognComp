```python
import numpy as np
import matplotlib.pyplot as plt
M = float(input("Enter the value of M: "))
x = np.linspace(-10,10,100)
y1 = M*x**2
y2 = np.sin(x)
plt.plot(x,y1,label="y = Mx^2", linestyle="-")
plt.plot(x,y2,label="y = Msin(x)", linestyle="-")
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of y = Mx^2 and y = sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import pandas as pd
data = {
    "Subject": ["Python", "DBMS", "Operating System", "Computer Networks", "Java"],
    "Score": [85, 78, 90, 82, 88]
}
df = pd.DataFrame(data)
sns.barplot(x="Subject", y="Score", data=df, hue="Subject", legend=False)
for i in range(len(df)):
    plt.text(i, df["Score"][i] + 1, str(df["Score"][i]),
             ha="center")
plt.title("Scores in Five Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")
plt.xticks(rotation=20)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

roll_no = int(input("Enter your roll number: "))

np.random.seed(roll_no)

data = np.random.randn(50)
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
cumulative_sum = np.cumsum(data)
plt.plot(cumulative_sum)
plt.title("Cumulative Sum")
plt.xlabel("Index")
plt.ylabel("Cumulative Sum")
plt.grid(True)
plt.subplot(2, 2, 2)

x = np.arange(50)
noise = np.random.randn(50)

plt.scatter(x, data + noise)
plt.title("Scatter Plot with Random Noise")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.subplot(2, 2, 3)

plt.plot(data)
plt.title("Original Random Data")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.subplot(2, 2, 4)

plt.hist(data)
plt.title("Distribution of Random Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)
print(df)

plt.figure(figsize=(10, 5))
sns.lineplot(x="month_number", y="total_profit", data=df, marker="o")
plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()

plt.figure(figsize=(12, 7))

sns.lineplot(x="month_number", y="facecream", data=df, marker="o", label="Face Cream")
sns.lineplot(x="month_number", y="facewash", data=df, marker="o", label="Face Wash")
sns.lineplot(x="month_number", y="toothpaste", data=df, marker="o", label="Toothpaste")
sns.lineplot(x="month_number", y="bathingsoap", data=df, marker="o", label="Bathing Soap")
sns.lineplot(x="month_number", y="shampoo", data=df, marker="o", label="Shampoo")
sns.lineplot(x="month_number", y="moisturizer", data=df, marker="o", label="Moisturizer")

plt.title("Product Sales Data")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")

plt.legend()
plt.grid(True)

plt.show()
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

total_sales = df[products].sum()
plt.figure(figsize=(10, 6))

sns.barplot(
    x=total_sales.index,
    y=total_sales.values,
    hue=total_sales.index,
    legend=False
)

plt.title("Total Sales of All Products")
plt.xlabel("Products")
plt.ylabel("Total Units Sold")

plt.xticks(rotation=20)
plt.grid(axis="y")

plt.show()
```
