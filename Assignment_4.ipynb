```python
import pandas as pd
roll_no = "1024170248"
fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"},
]
custom_entries = [
    {"question": "where is your office located", "answer": "Our main branch is in Building A.", "keywords": "location address office branch", "category": "general"},
    {"question": "how can i get an invoice", "answer": "Invoices are sent to your registered email.", "keywords": "invoice bill receipt receipt", "category": "billing"}
]
all_entries = fixed_entries + custom_entries
df = pd.DataFrame(all_entries)
print(df)
```

**Ques2**

```python
def score_query(query, dataframe):
    query_words = set(query.lower().split())
    scores = []

    for idx, row in dataframe.iterrows():
        text_to_check = set((row['question'] + " " + row['keywords']).lower().split())
        match_count = len(query_words.intersection(text_to_check))
        scores.append(match_count)

    temp_df = dataframe.copy()
    temp_df['score'] = scores

    ranked_df = temp_df[temp_df['score'] > 0].sort_values(by='score', ascending=False)
    return ranked_df

user_query = "how to pay fee"
print("\n--- Ranked Results for Query: 'how to pay fee' ---")
print(score_query(user_query, df)[['question', 'score', 'category']])
```

```python
def score_query_with_ties(query, dataframe):
    query_words = set(query.lower().split())
    scores = []

    for idx, row in dataframe.iterrows():
        text_to_check = set((row['question'] + " " + row['keywords']).lower().split())
        match_count = len(query_words.intersection(text_to_check))
        scores.append(match_count)

    dataframe_copy = dataframe.copy()
    dataframe_copy['score'] = scores

    max_score = max(scores)

    if max_score == 0:
        print("No relevant results found.")
        return pd.DataFrame()

    top_matches = dataframe_copy[dataframe_copy['score'] == max_score]

    if len(top_matches) > 1:
        print(f"Tie detected! Showing all {len(top_matches)} matches with top score {max_score}:")
    else:
        print(f"Single top match found (Score: {max_score}):")

    return top_matches[['question', 'answer', 'score']]

print(score_query_with_ties("fee", df))

print(score_query_with_ties("password reset", df))
```
