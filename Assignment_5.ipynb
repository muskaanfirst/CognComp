**Ques2**

```python
import numpy as np
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("a) Reversed array:", arr[::-1])
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("b) Array x: Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("b) Array y: Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

**Ques4**

```python
import numpy as np
Muskaan_Singhal = np.linspace(10, 100, 25)
print("Array:")
print(Muskaan_Singhal)
print("Dimensions:", Muskaan_Singhal.ndim)
print("Shape:", Muskaan_Singhal.shape)
print("Total elements:", Muskaan_Singhal.size)
print("Data type:", Muskaan_Singhal.dtype)
print("Total bytes:", Muskaan_Singhal.nbytes)
transpose_array = Muskaan_Singhal.reshape(1, 25)
print("Transpose using reshape:")
print(transpose_array)
print(Muskaan_Singhal.T)
```

**Ques5**

```python
import numpy as np
Muskaan_Singhal = np.array([[10, 20, 30, 40],
                            [50, 60, 70, 80],
                            [90, 15, 20, 35]])
print("Original array:")
print(Muskaan_Singhal)
print("Mean:", np.mean(Muskaan_Singhal))
print("Median:", np.median(Muskaan_Singhal))
print("Maximum:", np.max(Muskaan_Singhal))
print("Minimum:", np.min(Muskaan_Singhal))
print("Unique elements:", np.unique(Muskaan_Singhal))
resized_Muskaan_Singhal = np.resize(Muskaan_Singhal, (2, 3))

print("Resized array:")
print(resized_Muskaan_Singhal)
```
